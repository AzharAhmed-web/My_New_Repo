## Write a Program to demonstrate Aggregate, Vectorized, Lmbda(), Map(), Reduce() and FIlter() Functions in Numpy:


#### What is an Aggregation Function?

#### An aggregation function takes many values and turns them into one single value.

#### Think of it like this: you have 100 numbers, and you want just ONE number that represents them all (like the total, or the average).

````python
import numpy as np
a=np.array([10,20,30,40,50])
print("Array:",a)
print("Sum:",np.sum(a))
print("Mean:",np.mean(a))
print("Maximum:",np.max(a))
print("Minimum:",np.min(a))
print("Standard Deviation:",np.std(a))
````

### What is a Vectorized Operation in Numpy?

#### A vectorized operation means: doing an action on an entire array/column at once, instead of using a loop to go through each value one by one.

#### Pandas and NumPy are built to handle whole columns in one shot — this is much faster than writing a for loop.
### Program to demonstrate Vectorized Operation in Numpy:

````python
import numpy as np
marks=np.array([56,76,89,90,87,65])
print("Marks:",marks)
marks + 5
````

### Adding 10% to salary:

````python
salary=np.array([10000,25000,35000,40000,57000])
print("Salary before adding 10%:",salary)
salary=salary*1.1
print("Salary after adding 10%:",salary)
````

### Lambda Function in Numpy:

### What is a Lambda Function?

#### A lambda function is a small, one-line function that doesn't need a name. It's used when you need a quick function for a short task, without writing a full def function.

#### Also called: anonymous function (function without a name).
### Map Function with Numpy:

````python
a=np.array([1,2,3,4,5])
res=list(map(lambda x: x*x,a))
print("Square values:",res)
````

### Filter Function with Numpy:

````python
res1=list(filter(lambda x: x > 3,a))
print("Values that are greater than 3:",res1)
````

### Using lambda with Numpy:

````python
res2=list(map(lambda x: x+5,a))
print("After adding 5 to all values:",res2)
````

### Compare the execution time of a normal function (def) & a lambda function using the time module.

````python
import time
def square(x):
    return x*x
square_lambda=lambda x: x*x
start=time.time()
for i in range(10):
    square(i)
end=time.time()
print("Normal Function time:",end-start)
square_lambda(i)
print("Lambda Function:",end-start)
````

### Reduce Function with Numpy:

### What is reduce()?

#### reduce() takes a list of values and combines them into a single value, by repeatedly applying a function on pairs of elements, one after another.

#### It "reduces" the whole list down to one final result.
## Program to demonstrate reduce() function in NUMPY:

````python
import numpy as np
from functools import reduce
a1=np.array([10,20,30,40,50])
b=reduce(lambda x,y: x*y,a1)
print(b)
````

## GROUPING IN PANDAS:
### What is Grouping?

#### Grouping means: split your data into categories, then do some calculation on each category separately.

#### Think of it like sorting students into different sections, then finding the average marks for each section — not the whole class together.
## Program to demonstrate Grouping in PANDAS:

````python
import pandas as pd
data={ 
    "Name":["Azhar","Aftoz","Rayan","Tarun","Hafeez","Junaid"],
    "Branch":["CS","EC","AT","CS","CP","CS"],
    "Age":[17,18,19,17,19,20],
    "Marks":[98,28,45,54,67,36] 
    }
df=pd.DataFrame(data)
df
````

### Displaying the sum of the each branch marks:

````python
c=df.groupby("Branch").agg(Sum_Marks=("Marks","mean")).reset_index().round(2)
print(c)
````

### Displaying the Average(Mean) of Age and Marks:

````python
d=df.groupby("Branch").agg(Mean_Age=("Age","mean"),Mean_Marks=("Marks","mean")).reset_index().round(2)
print(d)
````

## TIME SERIES OPERATIONS IN PANDAS:
### What is Time Series Data?

### Time series data = data that has a date or time attached to each row.

### Example: daily sales, hourly temperature, monthly revenue.
## Program to demonstrate Monthly Sales in PANDAS:

````python
df2 = pd.DataFrame({
                    "Dates":["2024-01-01","2024-02-02","2024-01-03","2024-01-04","2024-02-05",
                             "2024-02-06","2024-03-07","2024-03-08","2024-01-09","2024-03-10"],
                    "Sales": [10,20,15,25,30,12,18,22,27,35],
                    "Month": ["JAN","FEB","JAN","JAN","FEB","FEB","MAR","MAR","JAN","MAR"],
                    "Profit": [10000,17000,21000,40000,27000,17000,30000,11000,23000,50000]
                   })
df2
````

### To Displaying the total profit of each month by grouping the months:

````python
tp=df2.groupby("Month")["Profit"].sum()
print(tp)
````
